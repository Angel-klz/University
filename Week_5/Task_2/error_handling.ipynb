<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Načítajte <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">sales_data_dirty.csv</span> tak, aby program vedel zrozumiteľne reagovať aj v prípade, že súbor neexistuje, je prázdny alebo sa ho nepodarí korektne parsovať.</p>

Príklad princípu:
<pre style="background: #f9fafb; padding: 14px; border-radius: 8px; overflow: auto; font-size: 14px; line-height: 1.5; font-family: ui-monospace, SFMono-Regular, Menlo, monospace; margin: 10px 0 14px;">try:

Pri CSV sa môžete stretnúť napríklad s <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">FileNotFoundError</span>,
<span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">EmptyDataError</span> alebo <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">ParserError</span>.
Vašou úlohou je vybrať vhodné výnimky a zobraziť zmysluplnú informáciu o tom, čo zlyhalo.
</div>


```python
import re
import warnings
import numpy as np
import pandas as pd

INPUT_FILE = "sales_data_dirty.csv"


def load_csv_safely(path, **kwargs):
    """Načíta CSV a zrozumiteľne ohlási typické chyby. Pri zlyhaní vráti None."""
    try:
        data = pd.read_csv(path, **kwargs)
    except FileNotFoundError:
        print(f"❌ Súbor '{path}' neexistuje. Skontrolujte cestu a pracovný priečinok.")
    except pd.errors.EmptyDataError:
        print(f"❌ Súbor '{path}' je prázdny (neobsahuje ani hlavičku).")
    except pd.errors.ParserError as e:
        print(f"❌ Súbor '{path}' sa nepodarilo spracovať – poškodená štruktúra CSV:\n   {e}")
    except UnicodeDecodeError:
        if "encoding" not in kwargs:
            print("⚠️ UTF-8 nefunguje, skúšam kódovanie cp1250 ...")
            return load_csv_safely(path, encoding="cp1250", **kwargs)
        print(f"❌ Súbor '{path}' sa nepodarilo dekódovať.")
    except PermissionError:
        print(f"❌ Nemáte oprávnenie čítať súbor '{path}' (je otvorený v inom programe?).")
    else:
        if data.empty:
            print(f"⚠️ Súbor '{path}' obsahuje iba hlavičku, bez dátových riadkov.")
            return None
        return data
    return None


df = load_csv_safely(INPUT_FILE, dtype={"OrderID": "string", "PhoneNumber": "string"})

if df is None:
    raise RuntimeError("Vstupný súbor sa nepodarilo načítať – ďalšie kroky nie je možné vykonať.")

if (df.columns != df.columns.str.strip()).any():
    print("ℹ️ Z názvov stĺpcov boli odstránené nadbytočné medzery.")
    df.columns = df.columns.str.strip()

print(f"Počet riadkov: {df.shape[0]}, počet stĺpcov: {df.shape[1]}")
print(df.head(5).to_string())
```

**Poznámka:** `load_csv_safely` odlišuje chýbajúci súbor (`FileNotFoundError`), prázdny súbor (`EmptyDataError`), poškodenú štruktúru (`ParserError`), zlé kódovanie a prístupové práva. Ak načítanie zlyhá, notebook sa zastaví s jasnou správou namiesto toho, aby pokračoval s `None`.

<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Pred ďalším spracovaním overte, či vstup obsahuje všetky stĺpce, ktoré pipeline očakáva. Ak niektorý chýba, spracovanie má skončiť s jasnou chybovou hláškou.</p>

<div
  style="margin-top: 12px; background-color: #f9fafb; border-left: 4px solid #60a5fa; padding: 12px; border-radius: 4px;">
  💡 <strong>Pomôcka:</strong> Pipeline často očakáva presnú schému. Môžete si preto vytvoriť
zoznam povinných názvov stĺpcov a porovnať ho s <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">df.columns</span>.
<br><br>
Princíp:
<pre style="background: #f9fafb; padding: 14px; border-radius: 8px; overflow: auto; font-size: 14px; line-height: 1.5; font-family: ui-monospace, SFMono-Regular, Menlo, monospace; margin: 10px 0 14px;">required = ["id", "date", "amount"]
# zistite, ktoré z required chýbajú v df.columns</pre>
Ak kritický stĺpec chýba, je vhodnejšie spracovanie zastaviť pomocou
<span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">raise ValueError(...)</span> než pokračovať s neúplnými dátami.
</div>


```python
REQUIRED_COLUMNS = [
    "OrderID", "Region", "Product", "Quantity", "SalesAmount",
    "Discount", "OrderDate", "DeliveryDate", "PhoneNumber",
]

missing_cols = [c for c in REQUIRED_COLUMNS if c not in df.columns]
extra_cols = [c for c in df.columns if c not in REQUIRED_COLUMNS]

if extra_cols:
    print("ℹ️ Nadbytočné stĺpce (pipeline ich nepoužije):", extra_cols)

if missing_cols:
    lower_map = {c.lower(): c for c in df.columns}
    hints = {c: lower_map[c.lower()] for c in missing_cols if c.lower() in lower_map}
    raise ValueError(
        f"Vo vstupe chýbajú povinné stĺpce: {missing_cols}. "
        f"Možné preklepy: {hints or 'žiadne'}. Dostupné stĺpce: {list(df.columns)}"
    )

print("✅ Všetky povinné stĺpce sú prítomné.")

raw_df = df.copy()

flags = pd.DataFrame(index=df.index)


def is_blank(series):
    """True, ak je hodnota NaN alebo iba medzery."""
    s = series.astype("string")
    return (s.isna() | s.str.strip().eq("")).fillna(False).astype(bool)


def blank_to_na(series):
    """Text bez okrajových medzier; prázdny reťazec sa mení na NA."""
    s = series.astype("string").str.strip()
    return s.mask(s.eq("").fillna(False))
```

**Poznámka:** Chýbajúci povinný stĺpec je kritická chyba schémy – pokračovať by znamenalo pracovať s neúplnými dátami, preto sa vyvoláva `ValueError`.

<h2 style="margin: 24px 0 16px; font-size: 20px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827; border-bottom: 2px solid #e5e7eb; padding-bottom: 6px;">💻 Úloha 4 – Validácia identifikátora objednávky</h2>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Stĺpec <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">OrderID</span> je kľúčový identifikátor záznamu. Overte jeho formát a unikátnosť.</p>
<ul style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.7; margin-top: 6px;">
<li>nájdite hodnoty, ktoré nezodpovedajú formátu <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">ORD</span> + 4 číslice,</li>
<li>nájdite všetky riadky, ktorých <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">OrderID</span> sa vyskytuje viac než raz,</li>
<li>duplicitné riadky zatiaľ nemažte, iba ich oddeľte na kontrolu.</li>
</ul>
<div
  style="margin-top: 12px; background-color: #f9fafb; border-left: 4px solid #60a5fa; padding: 12px; border-radius: 4px;">
  💡 <strong>Pomôcka:</strong> Identifikátor má dve samostatné požiadavky:
<strong>správny formát</strong> a <strong>unikátnosť</strong>.
<br><br>
Na formát sa hodí:
<pre style="background: #f9fafb; padding: 14px; border-radius: 8px; overflow: auto; font-size: 14px; line-height: 1.5; font-family: ui-monospace, SFMono-Regular, Menlo, monospace; margin: 10px 0 14px;">series.str.match(r"^ABC\d{3}$", na=False)</pre>
Toto je iba príklad iného formátu. Pre vaše <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">OrderID</span> si zostavte vlastný regex.
<br><br>
Na duplicity sa hodí:
<pre style="background: #f9fafb; padding: 14px; border-radius: 8px; overflow: auto; font-size: 14px; line-height: 1.5; font-family: ui-monospace, SFMono-Regular, Menlo, monospace; margin: 10px 0 14px;">df.duplicated("id", keep=False)</pre>
Najprv si duplicitné riadky zobrazte. Neodstraňujte ich automaticky.
</div>


```python
oid = blank_to_na(df["OrderID"]).str.upper()     

format_ok = oid.str.match(r"^ORD\d{4}$", na=False).fillna(False).astype(bool)

flags["missing_orderid"] = oid.isna().astype(bool)
flags["invalid_orderid_format"] = (oid.notna() & ~format_ok).astype(bool)
flags["duplicate_orderid"] = (oid.notna() & oid.duplicated(keep=False)).astype(bool)

df["OrderID_clean"] = oid.where(format_ok)

print("Chýbajúce OrderID:", flags["missing_orderid"].sum())

print("\n--- OrderID v nesprávnom formáte (nie ORD + 4 číslice) ---")
print(df.loc[flags["invalid_orderid_format"], ["OrderID"]].to_string())

print("\n--- Riadky s duplicitným OrderID (zatiaľ ich nemažeme) ---")
dups = df.loc[flags["duplicate_orderid"]].assign(_key=oid).sort_values(["_key"]).drop(columns="_key")
print(dups.to_string())
print(f"\nRiadkov v duplicitných skupinách: {len(dups)}, "
      f"unikátnych duplicitných ID: {oid[flags['duplicate_orderid']].nunique()}")
```

**Rozhodnutie:** orezanie medzier a prevod na veľké písmená (`ord0001` → `ORD0001`) je jednoznačná oprava. Chýbajúci alebo zle formátovaný identifikátor sa neopravuje. Pri duplicitách nevieme, ktorý riadok je správny, preto sa všetky riadky s opakovaným ID neskôr zamietnu na kontrolu.

Pri hodnote, ktorej význam si neviete jednoznačne odvodiť, ju radšej označte ako problém namiesto toho, aby ste ju „uhádli“.
</div>


```python
from difflib import SequenceMatcher

ALLOWED_REGIONS = ["North", "South", "East", "West"]
ALLOWED_PRODUCTS = ["Laptop", "Phone", "Tablet", "Monitor"]

REGION_ALIASES = {"n": "North", "s": "South", "e": "East", "w": "West"}
PRODUCT_ALIASES = {}  


def normalize_text(series):
    """strip + lower + odstránenie interpunkcie a zdvojených medzier."""
    s = series.astype("string").str.strip().str.lower()
    s = s.str.replace(r"[.,;:!?'\"()\[\]/\\_-]", "", regex=True).str.replace(r"\s+", " ", regex=True).str.strip()
    return s.mask(s.eq("").fillna(False))


def match_category(value, allowed, aliases, cutoff=0.75, margin=0.10):
    """Presná zhoda -> alias -> zjavný preklep. Inak NA (nehádame)."""
    if pd.isna(value):
        return pd.NA
    lookup = {a.lower(): a for a in allowed}
    if value in lookup:
        return lookup[value]
    if value in aliases:
        return aliases[value]
    scored = sorted(((SequenceMatcher(None, value, k).ratio(), k) for k in lookup), reverse=True)
    best = scored[0]
    second = scored[1][0] if len(scored) > 1 else 0.0
    if best[0] >= cutoff and (best[0] - second) >= margin:   
        return lookup[best[1]]
    return pd.NA


def clean_category(series, allowed, aliases):
    norm = normalize_text(series)
    mapping = {v: match_category(v, allowed, aliases) for v in norm.dropna().unique()}
    return norm.map(mapping).astype("string")


for col, allowed, aliases in [("Region", ALLOWED_REGIONS, REGION_ALIASES),
                              ("Product", ALLOWED_PRODUCTS, PRODUCT_ALIASES)]:
    cleaned = clean_category(df[col], allowed, aliases)
    df[f"{col}_clean"] = cleaned

    blank = is_blank(df[col])
    flags[f"missing_{col.lower()}"] = blank
    flags[f"invalid_{col.lower()}"] = (~blank & cleaned.isna()).astype(bool)

    print(f"\n=== {col}: pôvodná -> vyčistená hodnota ===")
    audit = (pd.DataFrame({"raw": df[col], "clean": cleaned})
             .value_counts(dropna=False).rename("count").reset_index())
    print(audit.to_string(index=False))

    unmapped = df.loc[flags[f"invalid_{col.lower()}"], col].unique()
    print(f"Nezmapované hodnoty (označené ako neplatné, nehádame): {list(unmapped)}")

    # overenie: ostali iba povolené kategórie
    remaining = set(cleaned.dropna().unique())
    assert remaining <= set(allowed), f"{col}: nepovolené hodnoty {remaining - set(allowed)}"
    print(f"✅ {col}_clean obsahuje iba: {sorted(remaining)} (+ {cleaned.isna().sum()} NA)")
```

**Rozhodnutie:** normalizácia (medzery, veľkosť písmen, interpunkcia) je oddelená od mapovania významu. Skratky `N/S/E/W` sa mapujú explicitne, preklepy iba ak je najbližšia kategória jednoznačná. Čokoľvek iné sa neuhádne, ale označí ako `invalid_*` a záznam sa neskôr zamietne.

<h2 style="margin: 24px 0 16px; font-size: 20px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827; border-bottom: 2px solid #e5e7eb; padding-bottom: 6px;">💻 Úloha 6 – Číselné hodnoty a konverzné chyby</h2>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Stĺpce <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">Quantity</span>, <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">SalesAmount</span> a <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">Discount</span> obsahujú hodnoty, ktoré nie sú uložené konzistentne.</p>
<ul style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.7; margin-top: 6px;">
<li>identifikujte hodnoty, ktoré sa nedajú priamo skonvertovať na číslo, odstráňte formátovacie znaky, pri ktorých je význam jednoznačný (napr. mena alebo oddeľovač tisícov), konverziu vykonajte s <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">errors='coerce'</span> a následne odlíšte pôvodne chýbajúce hodnoty od hodnôt, ktoré sa stali <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">NaN</span> kvôli chybe konverzie,</li>
<li>overte business pravidlá pre rozsah hodnôt.</li>
</ul>
<div
  style="margin-top: 12px; background-color: #f9fafb; border-left: 4px solid #60a5fa; padding: 12px; border-radius: 4px;">
  💡 <strong>Pomôcka:</strong> Číselný stĺpec môže obsahovať napríklad medzery,
symbol meny alebo textové hodnoty. Najprv odstráňte iba tie znaky, pri ktorých je význam jasný.
<br><br>
Príklad:
<pre style="background: #f9fafb; padding: 14px; border-radius: 8px; overflow: auto; font-size: 14px; line-height: 1.5; font-family: ui-monospace, SFMono-Regular, Menlo, monospace; margin: 10px 0 14px;">cleaned = series.str.replace("$", "", regex=False)
numeric = pd.to_numeric(cleaned, errors="coerce")</pre>
Po konverzii hľadajte <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">NaN</span> a následne overte business pravidlá,
napríklad kladnosť alebo povolený interval.
<br><br>
Pozor: <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">NaN</span> po konverzii môže znamenať niečo iné než pôvodne chýbajúca hodnota.
</div>


```python
CURRENCY_RE = r"(?i)(€|\$|£|eur|usd|gbp|czk|kč|kc)"


def normalize_number_string(x):
    """Jednoznačné formáty tisícov/desatinnej časti -> text s desatinnou bodkou."""
    if pd.isna(x):
        return pd.NA
    if re.fullmatch(r"[+-]?\d{1,3}(,\d{3})+(\.\d+)?", x):         
        return x.replace(",", "")
    if re.fullmatch(r"[+-]?\d{1,3}(\.\d{3})+,\d+", x):            
        return x.replace(".", "").replace(",", ".")
    if re.fullmatch(r"[+-]?\d+,\d{1,2}", x):                     
        return x.replace(",", ".")
    return x


def to_number(series, percent=False):
    """Vráti (číslo, bolo_pôvodne_prázdne, chyba_konverzie)."""
    blank = is_blank(series)
    t = blank_to_na(series).str.replace(CURRENCY_RE, "", regex=True)
    if percent:
        t = t.str.replace("%", "", regex=False)
    t = t.str.replace("[\\s\u00a0\u202f]", "", regex=True)          
    t = t.mask(t.eq("").fillna(False))
    mapping = {v: normalize_number_string(v) for v in t.dropna().unique()}
    t = t.map(mapping).astype("string")
    num = pd.to_numeric(t, errors="coerce").astype("Float64").to_numpy(dtype="float64", na_value=np.nan)
    num = pd.Series(num, index=series.index)
    num = num.where(np.isfinite(num))                                 
    conv_error = (~blank & num.isna()).astype(bool)                 
    return num, blank, conv_error


qty, qty_blank, qty_conv = to_number(df["Quantity"])
qty_ok = (qty > 0) & (qty == np.floor(qty))
flags["missing_quantity"] = qty_blank
flags["conversion_error_quantity"] = qty_conv
flags["invalid_quantity_rule"] = (qty.notna() & ~qty_ok).astype(bool)
df["Quantity_clean"] = qty.where(qty_ok).astype("Int64")

amt, amt_blank, amt_conv = to_number(df["SalesAmount"])
amt_ok = amt > 0
flags["missing_salesamount"] = amt_blank
flags["conversion_error_salesamount"] = amt_conv
flags["invalid_salesamount_rule"] = (amt.notna() & ~amt_ok).astype(bool)
df["SalesAmount_clean"] = amt.where(amt_ok)

disc, disc_blank, disc_conv = to_number(df["Discount"], percent=True)
disc_ok = disc.between(0, 100)
flags["missing_discount"] = disc_blank
flags["conversion_error_discount"] = disc_conv
flags["invalid_discount_rule"] = (disc.notna() & ~disc_ok).astype(bool)
df["Discount_clean"] = disc.where(disc_ok)

for col, blank, conv, ok, num in [("Quantity", qty_blank, qty_conv, qty_ok, qty),
                                  ("SalesAmount", amt_blank, amt_conv, amt_ok, amt),
                                  ("Discount", disc_blank, disc_conv, disc_ok, disc)]:
    print(f"\n=== {col} ===")
    print("pôvodne chýbajúce:       ", int(blank.sum()))
    print("chyba konverzie (text):  ", int(conv.sum()), "->", list(df.loc[conv, col].unique()))
    print("porušenie business pravidla:", int((num.notna() & ~ok).sum()),
          "->", num[num.notna() & ~ok].unique().tolist())
```

**Rozhodnutie:** odstraňujú sa iba znaky s jasným významom (mena, `%`, medzery, tisícové oddeľovače). `errors="coerce"` mení nečíselný text na `NaN`, no vďaka porovnaniu s pôvodnou hodnotou rozlišujeme **pôvodne chýbajúce** (`missing_*`) od **chyby konverzie** (`conversion_error_*`) a od **porušenia pravidla** (`invalid_*_rule`). Nič sa nenahrádza nulou. Nejednoznačný zápis `1,200` sa berie ako tisícový oddeľovač.

<h2 style="margin: 24px 0 16px; font-size: 20px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827; border-bottom: 2px solid #e5e7eb; padding-bottom: 6px;">💻 Úloha 7 – Dátumy a logická konzistencia</h2>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Skonvertujte <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">OrderDate</span> a <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">DeliveryDate</span> na dátumový typ a skontrolujte nielen formát, ale aj vzájomnú logiku.</p>
<ul style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.7; margin-top: 6px;">
<li>neplatné dátumy preveďte na <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">NaT</span>,</li>
<li>nájdite záznamy, v ktorých sa dátum nepodarilo skonvertovať,nájdite objednávky, pri ktorých je <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">DeliveryDate</span> skorší než <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">OrderDate</span>.</li>
</ul>
<div
  style="margin-top: 12px; background-color: #f9fafb; border-left: 4px solid #60a5fa; padding: 12px; border-radius: 4px;">
  💡 <strong>Pomôcka:</strong> Prvý krok je bezpečná konverzia:
<pre style="background: #f9fafb; padding: 14px; border-radius: 8px; overflow: auto; font-size: 14px; line-height: 1.5; font-family: ui-monospace, SFMono-Regular, Menlo, monospace; margin: 10px 0 14px;">pd.to_datetime(series, errors="coerce")</pre>
Neplatné hodnoty sa zmenia na <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">NaT</span>.
Potom však ešte musíte skontrolovať logiku medzi dvoma dátumami.
<br><br>
Príklad logickej kontroly:
<pre style="background: #f9fafb; padding: 14px; border-radius: 8px; overflow: auto; font-size: 14px; line-height: 1.5; font-family: ui-monospace, SFMono-Regular, Menlo, monospace; margin: 10px 0 14px;">end_date < start_date</pre>
Takýto záznam môže obsahovať dva úplne platné dátumy, ale ich vzájomný vzťah je chybný.
</div>


```python
DAYFIRST = False  


def parse_dates(series):
    blank = is_blank(series)
    text = series.astype("string").str.strip()
    dotted = text.str.match(r"^\d{1,2}\.\d{1,2}\.\d{4}$", na=False).fillna(False).astype(bool)
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        parsed = pd.to_datetime(text.mask(dotted), errors="coerce", format="mixed", dayfirst=DAYFIRST)
        parsed_dotted = pd.to_datetime(text.where(dotted), errors="coerce", format="mixed", dayfirst=True)
    parsed = parsed.fillna(parsed_dotted)
    conv_error = (~blank & parsed.isna()).astype(bool)
    return parsed, blank, conv_error


order_dt, order_blank, order_conv = parse_dates(df["OrderDate"])
deliv_dt, deliv_blank, deliv_conv = parse_dates(df["DeliveryDate"])

flags["missing_orderdate"] = order_blank
flags["invalid_orderdate"] = order_conv
flags["missing_deliverydate"] = deliv_blank
flags["invalid_deliverydate"] = deliv_conv
flags["delivery_before_order"] = (order_dt.notna() & deliv_dt.notna() & (deliv_dt < order_dt)).astype(bool)

df["OrderDate_clean"] = order_dt
df["DeliveryDate_clean"] = deliv_dt

print("--- Dátumy, ktoré sa nepodarilo skonvertovať (NaT) ---")
print(df.loc[order_conv | deliv_conv, ["OrderID", "OrderDate", "DeliveryDate"]].to_string())

print("\n--- DeliveryDate skorší než OrderDate ---")
print(df.loc[flags["delivery_before_order"], ["OrderID", "OrderDate_clean", "DeliveryDate_clean"]].to_string())
```

**Poznámka:** záznam môže mať dva platné dátumy, ktorých vzťah je chybný (dodanie pred objednávkou). Logická kontrola sa preto robí iba tam, kde sú obe hodnoty skonvertované. Rovnaký deň objednávky a dodania je povolený.

<h2 style="margin: 24px 0 16px; font-size: 20px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827; border-bottom: 2px solid #e5e7eb; padding-bottom: 6px;">💻 Úloha 8 – Validácia telefónneho čísla</h2>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Telefónne čísla sú v súbore zapísané v rôznych formátoch. Stĺpec je nepovinný, ale vyplnená hodnota musí byť validná.</p>
<ul style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.7; margin-top: 6px;">
<li>odstráňte medzery, zátvorky, pomlčky a znak <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">+</span>,</li>
<li>ponechajte iba číslice, za platné považujte hodnoty s 10 až 15 číslicami, neplatné, ale vyplnené telefónne čísla označte samostatným validačným príznakom.</li>
</ul>
<div
  style="margin-top: 12px; background-color: #f9fafb; border-left: 4px solid #60a5fa; padding: 12px; border-radius: 4px;">
  💡 <strong>Pomôcka:</strong> Telefónne čísla môžu mať rôzne formátovanie.
Najprv preto vytvorte pomocnú hodnotu, v ktorej zostanú iba číslice.
<pre style="background: #f9fafb; padding: 14px; border-radius: 8px; overflow: auto; font-size: 14px; line-height: 1.5; font-family: ui-monospace, SFMono-Regular, Menlo, monospace; margin: 10px 0 14px;">digits = series.str.replace(r"\D", "", regex=True)</pre>
Potom môžete kontrolovať dĺžku pomocou <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">str.len()</span>.
<br><br>
Rozlišujte medzi <strong>chýbajúcou hodnotou</strong> a <strong>vyplnenou, ale neplatnou hodnotou</strong>.
Keď je telefón nepovinný, prázdna hodnota nemusí byť chyba.
</div>


```python
phone_blank = is_blank(df["PhoneNumber"])

digits = (df["PhoneNumber"].astype("string")
          .str.replace(r"\.0$", "", regex=True)
          .str.replace(r"\D", "", regex=True))
digit_len = digits.str.len()

phone_valid = (~phone_blank & digit_len.between(10, 15)).fillna(False).astype(bool)

flags["invalid_phone"] = (~phone_blank & ~phone_valid).astype(bool)      
df["PhoneNumber_clean"] = digits.where(phone_valid)                      

print("Prázdne (povolené):        ", int(phone_blank.sum()))
print("Vyplnené a platné:         ", int(phone_valid.sum()))
print("Vyplnené, ale neplatné:    ", int(flags["invalid_phone"].sum()))
print("\n--- Neplatné telefónne čísla ---")
print(df.loc[flags["invalid_phone"], ["OrderID", "PhoneNumber"]].assign(digits=digits).to_string())
```

**Rozhodnutie:** prázdny telefón nie je chyba (pole je nepovinné). Vyplnené číslo mimo 10–15 číslic dostane samostatný príznak `invalid_phone`.

<h2 style="margin: 24px 0 16px; font-size: 20px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827; border-bottom: 2px solid #e5e7eb; padding-bottom: 6px;">💻 Úloha 9 – Vytvorenie validačného reportu</h2>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Namiesto toho, aby boli jednotlivé kontroly roztrúsené po notebooku, vytvorte jeden prehľad, ktorý ukáže počet nájdených problémov podľa typu.</p>
<ul style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.7; margin-top: 6px;">
<li>vytvorte validačné príznaky pre jednotlivé pravidlá, zostavte z nich tabuľku <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">validation_report</span>,</li>
<li>pri každom pravidle zobrazte počet chybných riadkov.</li>
</ul>
<div
  style="margin-top: 12px; background-color: #f9fafb; border-left: 4px solid #60a5fa; padding: 12px; border-radius: 4px;">
  💡 <strong>Pomôcka:</strong> Dobrá validačná pipeline si pri každom pravidle vytvorí boolean príznak:
<span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">True</span> = problém, <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">False</span> = v poriadku.
<br><br>
Napríklad všeobecne:
<pre style="background: #f9fafb; padding: 14px; border-radius: 8px; overflow: auto; font-size: 14px; line-height: 1.5; font-family: ui-monospace, SFMono-Regular, Menlo, monospace; margin: 10px 0 14px;">flags["invalid_amount"] = ...
flags["invalid_date"] = ...</pre>
Keď máte pravidlá pokope, <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">sum()</span> vám dá počet chýb pre každý typ.
Takto získate report, ktorý sa dá neskôr logovať alebo monitorovať.
</div>


```python
RULES = {
    "missing_orderid":              ("OrderID chýba", "critical"),
    "invalid_orderid_format":       ("OrderID nemá formát ORDdddd", "critical"),
    "duplicate_orderid":            ("OrderID sa opakuje", "critical"),
    "missing_region":               ("Region chýba", "critical"),
    "invalid_region":               ("Region nie je North/South/East/West", "critical"),
    "missing_product":              ("Product chýba", "critical"),
    "invalid_product":              ("Product nie je v povolenej množine", "critical"),
    "missing_quantity":             ("Quantity chýba", "critical"),
    "conversion_error_quantity":    ("Quantity nie je číslo", "critical"),
    "invalid_quantity_rule":        ("Quantity nie je kladné celé číslo", "critical"),
    "missing_salesamount":          ("SalesAmount chýba", "critical"),
    "conversion_error_salesamount": ("SalesAmount nie je číslo", "critical"),
    "invalid_salesamount_rule":     ("SalesAmount nie je kladný", "critical"),
    "missing_discount":             ("Discount chýba (nahradiť nulou by bolo hádanie)", "warning"),
    "conversion_error_discount":    ("Discount nie je číslo", "critical"),
    "invalid_discount_rule":        ("Discount mimo intervalu 0–100", "critical"),
    "missing_orderdate":            ("OrderDate chýba", "critical"),
    "invalid_orderdate":            ("OrderDate nie je platný dátum", "critical"),
    "missing_deliverydate":         ("DeliveryDate chýba", "critical"),
    "invalid_deliverydate":         ("DeliveryDate nie je platný dátum", "critical"),
    "delivery_before_order":        ("DeliveryDate je skôr než OrderDate", "critical"),
    "invalid_phone":                ("PhoneNumber vyplnené, ale nemá 10–15 číslic", "warning"),
}

assert set(flags.columns) == set(RULES), set(flags.columns) ^ set(RULES)
flags = flags[list(RULES)].astype(bool)

validation_report = pd.DataFrame({
    "rule": list(RULES),
    "description": [RULES[r][0] for r in RULES],
    "severity": [RULES[r][1] for r in RULES],
    "failed_rows": [int(flags[r].sum()) for r in RULES],
})
validation_report["pct_of_rows"] = (validation_report["failed_rows"] / len(df) * 100).round(2)

print(validation_report.to_string(index=False))
print(f"\nCelkový počet riadkov: {len(df)}")
print(f"Riadky aspoň s jednou chybou: {int(flags.any(axis=1).sum())}")
```

**Poznámka:** každé pravidlo má vlastný boolean príznak (`True` = problém), takže `flags.sum()` dáva počet chýb podľa typu a report sa dá neskôr logovať alebo monitorovať. Jeden riadok môže porušovať viac pravidiel naraz.

<h2 style="margin: 24px 0 16px; font-size: 20px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827; border-bottom: 2px solid #e5e7eb; padding-bottom: 6px;">💻 Úloha 10 – Oddelenie validných a problémových záznamov</h2>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Nie všetky chyby majú rovnakú závažnosť. Navrhnite, ktoré chyby považujete za kritické pre ďalšie spracovanie a ktoré môžu zostať iba ako upozornenie.</p>
<ul style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.7; margin-top: 6px;">
<li>vytvorte masku kritických chýb, rozdeľte dáta na <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">clean_data</span> a <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">rejected_records</span>,</li>
<li>pri zamietnutých záznamoch zachovajte informáciu o dôvode zamietnutia.</li>
</ul>
<div
  style="margin-top: 12px; background-color: #f9fafb; border-left: 4px solid #60a5fa; padding: 12px; border-radius: 4px;">
  💡 <strong>Pomôcka:</strong> Ak má každý validačný test vlastný boolean stĺpec,
môžete sa na úroveň celého riadku opýtať:
<em>porušil tento riadok aspoň jedno kritické pravidlo?</em>
<br><br>
Na to sa hodí:
<pre style="background: #f9fafb; padding: 14px; border-radius: 8px; overflow: auto; font-size: 14px; line-height: 1.5; font-family: ui-monospace, SFMono-Regular, Menlo, monospace; margin: 10px 0 14px;">flags[critical_rules].any(axis=1)</pre>
Výsledkom je maska, ktorou môžete rozdeliť dáta na validné a problémové.
Pri problémových riadkoch je užitočné ponechať aj informáciu, ktoré pravidlá zlyhali.
</div>


```python
critical_rules = [r for r, (_, sev) in RULES.items() if sev == "critical"]
warning_rules = [r for r, (_, sev) in RULES.items() if sev == "warning"]


def join_flags(frame):
    """Pre každý riadok spojí názvy porušených pravidiel do textu."""
    return pd.Series([", ".join(frame.columns[row]) for row in frame.to_numpy()],
                     index=frame.index, dtype="object")


critical_mask = flags[critical_rules].any(axis=1)

clean_data = df.loc[~critical_mask].copy()

rejected_records = raw_df.loc[critical_mask].copy()           
rejected_records.insert(0, "source_row", rejected_records.index + 2)   
rejected_records["rejection_reasons"] = join_flags(flags.loc[critical_mask, critical_rules])
rejected_records["warnings"] = join_flags(flags.loc[critical_mask, warning_rules])

print(f"Všetky riadky:     {len(df)}")
print(f"clean_data:        {len(clean_data)}")
print(f"rejected_records:  {len(rejected_records)}")
assert len(clean_data) + len(rejected_records) == len(df)

print("\n--- Dôvody zamietnutia (počet riadkov) ---")
print(flags.loc[critical_mask, critical_rules].sum().loc[lambda s: s > 0].sort_values(ascending=False).to_string())

print("\n--- Ukážka zamietnutých záznamov ---")
print(rejected_records[["source_row", "OrderID", "rejection_reasons"]].head(10).to_string(index=False))
```

**Návrh závažnosti:** *kritické* sú chyby, ktoré robia záznam nepoužiteľným alebo nespoľahlivým (kľúč, kategórie, množstvo, suma, dátumy, zľava mimo rozsahu). *Upozornenia* sú iba chýbajúca zľava a neplatný nepovinný telefón – záznam zostáva, no telefón sa vo finálnych dátach nastaví na prázdny.

<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">V dátach, ktoré prešli kritickou validáciou, nahraďte pôvodné nekonzistentné stĺpce vyčistenými verziami a ponechajte iba finálnu schému vhodnú na ďalšie spracovanie.</p>

<h2 style="margin: 24px 0 16px; font-size: 20px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827; border-bottom: 2px solid #e5e7eb; padding-bottom: 6px;">💻 Úloha 12 – Opätovná validácia a error handling</h2>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Po čistení vykonajte validačné pravidlá ešte raz. Ak kritické chyby stále zostali, pipeline nemá pokračovať ako keby bolo všetko v poriadku.</p>
<ul style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.7; margin-top: 6px;">
<li>vytvorte vlastnú výnimku <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">DataValidationError</span> alebo použite vhodnú existujúcu výnimku,</li>
<li>overte unikátnosť <span style="background-color: #e5e7eb; color: #111827; font-family: monospace; padding: 2px 5px; border-radius: 4px;">OrderID</span>,</li>
<li>overte rozsahy číselných hodnôt a logiku dátumov, pri zlyhaní vyvolajte výnimku s informatívnou správou.</li>
</ul>
<div
  style="margin-top: 12px; background-color: #f9fafb; border-left: 4px solid #60a5fa; padding: 12px; border-radius: 4px;">
  💡 <strong>Pomôcka:</strong> Po čistení je vhodné najdôležitejšie pravidlá spustiť ešte raz.
Ak kritická podmienka stále neplatí, spracovanie sa má zastaviť.
<br><br>
Vlastná výnimka môže mať tvar:
<pre style="background: #f9fafb; padding: 14px; border-radius: 8px; overflow: auto; font-size: 14px; line-height: 1.5; font-family: ui-monospace, SFMono-Regular, Menlo, monospace; margin: 10px 0 14px;">class MyValidationError(Exception):
    pass</pre>
A pri zistení problému ju môžete vyvolať pomocou:
<pre style="background: #f9fafb; padding: 14px; border-radius: 8px; overflow: auto; font-size: 14px; line-height: 1.5; font-family: ui-monospace, SFMono-Regular, Menlo, monospace; margin: 10px 0 14px;">raise MyValidationError("...")</pre>
Nejde o to vyhadzovať výnimku pri každom NULL-e. Výnimka je vhodná vtedy, keď ďalšie spracovanie už nie je bezpečné.
</div>


```python
class DataValidationError(Exception):
    """Kritická chyba kvality dát – ďalšie spracovanie nie je bezpečné."""


def validate_final(data):
    problems = []

    if list(data.columns) != list(COLUMN_MAP.values()):
        problems.append(f"Nesprávna schéma: {list(data.columns)}")
        return problems
    if data.empty:
        problems.append("Po validácii nezostal žiadny platný záznam.")
        return problems

    if data["OrderID"].isna().any():
        problems.append("OrderID obsahuje prázdne hodnoty.")
    bad_fmt = ~data["OrderID"].str.match(r"^ORD\d{4}$", na=False).fillna(False).astype(bool)
    if bad_fmt.any():
        problems.append(f"OrderID so zlým formátom: {int(bad_fmt.sum())} riadkov.")
    if data["OrderID"].duplicated().any():
        problems.append(f"Duplicitné OrderID: {data.loc[data['OrderID'].duplicated(), 'OrderID'].tolist()}")

    if not set(data["Region"].astype(str)) <= set(ALLOWED_REGIONS) or data["Region"].isna().any():
        problems.append("Region obsahuje nepovolené alebo prázdne hodnoty.")
    if not set(data["Product"].astype(str)) <= set(ALLOWED_PRODUCTS) or data["Product"].isna().any():
        problems.append("Product obsahuje nepovolené alebo prázdne hodnoty.")

    if data["Quantity"].isna().any() or (data["Quantity"] <= 0).any():
        problems.append("Quantity musí byť kladné celé číslo.")
    if data["SalesAmount"].isna().any() or (data["SalesAmount"] <= 0).any():
        problems.append("SalesAmount musí byť kladná hodnota.")
    if (~data["Discount"].dropna().between(0, 100)).any():
        problems.append("Discount je mimo intervalu 0–100.")

    if data[["OrderDate", "DeliveryDate"]].isna().any().any():
        problems.append("Dátumy obsahujú prázdne hodnoty (NaT).")
    if (data["DeliveryDate"] < data["OrderDate"]).any():
        problems.append("Existujú objednávky s DeliveryDate skôr než OrderDate.")

    phone_len = data["PhoneNumber"].dropna().str.len()
    if (~phone_len.between(10, 15)).any():
        problems.append("PhoneNumber má vyplnenú hodnotu mimo 10–15 číslic.")

    return problems


def assert_valid(data):
    problems = validate_final(data)
    if problems:
        raise DataValidationError("Validácia zlyhala:\n - " + "\n - ".join(problems))
    print(f"✅ Opätovná validácia prešla ({len(data)} riadkov) – pipeline môže pokračovať.")


assert_valid(final_df)

broken = final_df.copy()
broken.loc[broken.index[0], "Quantity"] = -5
try:
    assert_valid(broken)
except DataValidationError as e:
    print("\nOčakávaná chyba na zámerne pokazených dátach:\n", e)
```

**Poznámka:** výnimka sa nevyvoláva pri každom `NULL`, ale iba tam, kde by ďalšie spracovanie bolo nebezpečné (porušený kľúč, rozsah, logika dátumov, schéma). Prázdny `Discount` či `PhoneNumber` sú povolené.
